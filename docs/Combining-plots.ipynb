## 2D combinations

### Combining plots

```raku
#%js
js-d3-combine(
    js-d3-layer-line([[1, 2], [2, 5]], :color<steelblue>, :legend<Observed>),
    js-d3-layer-points([[1, 3], [2, 4]], :color<tomato>, :legend<Forecast>),
    js-d3-layer-labels([%(x => 2, y => 4, label => 'Latest')]),
    :title<Combined view>,
    :x-axis-label<Time>,
    :y-axis-label<Value>,
    :$background,
    :$title-color
    );
```

### Combining a chart with plots

```raku
#% js
js-d3-combine(
    js-d3-layer-histogram([2, 3, 3, 4, 6, 8], :color<tomato>, :legend('Sample B'), :number-of-bins(4)),
    js-d3-layer-line([[1, 1], [8, 3]], :color<Salmon>, :legend<Trend>),
    js-d3-layer-points([[2, 2], [6, 4]], :color<Gold>, :legend<Observations>),
    :title<Histogram line points>,
    :x-axis-label<Value>,
    :y-axis-label<Count>,
    :$background,
    :$title-color
)
```

### Combining charts

```raku
#% js
js-d3-combine(
    js-d3-layer-histogram([1, 2, 2, 3, 5, 8], :color<steelblue>, :legend('Sample A'), :number-of-bins(4)),
    js-d3-layer-histogram([2, 3, 3, 4, 6, 8], :color<tomato>, :legend('Sample B'), :number-of-bins(4)),
)
```
