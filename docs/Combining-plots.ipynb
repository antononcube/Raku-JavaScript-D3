## 2D combinations

### Combining plots

```raku
#%js
js-d3-combine(
    js-d3-layer-line([[1, 2], [2, 5]], :color<steelblue>, :legend<Observed>),
    js-d3-layer-points([[1, 3], [2, 4]], :color<tomato>, :legend<Forecast>),
    js-d3-layer-labels([%(x => 2, y => 4, label => 'Latest')]),
    :title<Combined view>,
    :x-axis-label<Time>,
    :y-axis-label<Value>,
    :$background,
    :$title-color
    );
```

### Combining a chart with plots

```raku
#% js
js-d3-combine(
    js-d3-layer-histogram([2, 3, 3, 4, 6, 8], :color<tomato>, :legend('Sample B'), :number-of-bins(4)),
    js-d3-layer-line([[1, 1], [8, 3]], :color<LightBlue>, :legend<Trend>),
    js-d3-layer-points([[2, 2], [6, 4]], :color<Gold>, :legend<Observations>),
    :title<Histogram line points>,
    :x-axis-label<Value>,
    :y-axis-label<Count>,
    :$background,
    :$title-color
)
```

### Combining charts

```raku
#%js
my @sample-a = random-variate(NormalDistribution.new(10,5), 300);
my @sample-b = random-variate(NormalDistribution.new(25,10), 300);
js-d3-combine(
    js-d3-layer-histogram(@sample-a, :color<steelblue>, :legend<A>),
    js-d3-layer-histogram(@sample-b, :color<tomato>, :legend<B>, :opacity(0.55)),
    :title<Sample comparison>,
    :x-axis-label<Value>,
    :y-axis-label<Count>,
    :$background,
    :$title-color
)
```

```raku
#% js
js-d3-combine(
    js-d3-layer-box-whisker(@sample-a, :x(1), :color<steelBlue>, :legend<A>, :horizontal),
    js-d3-layer-box-whisker(@sample-b, :x(2), :color<tomato>, :legend<B>, :horizontal),
    js-d3-layer-box-whisker(3 <<*>> @sample-b, :x(3), :color<pink>, :legend<C>, :horizontal),
    js-d3-layer-box-whisker(1.2 <<*>> @sample-b, :x(4), :color<lightblue>, :legend<D>, :horizontal),
    :x-axis-label<Group position>,
    :y-axis-label<Value>,
    :$background,
    :!axes,
    :200height
)
```
