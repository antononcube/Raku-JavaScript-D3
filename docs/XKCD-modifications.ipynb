## Setup

Here we use a JavaScript cell that allows the visualization of with [D3.js](https://d3js.org) in Jupyter notebooks:


```raku
%% javascript
require.config({
  paths: {
    d3: "https://d3js.org/d3.v7.min",
    d3_3d: "https://unpkg.com/d3-3d@2.0.2/build/d3-3d"
  }
});

require(["d3", "d3_3d"], function(d3, d3_3d) {
  window.d3 = d3;
  window.d33d = d3_3d || window.d33d || {};
  console.log("d3:", d3.version, "d3_3d:", Object.keys(window.d33d));
});
```

```raku
my $background = '#1A1A1A';
my $stroke-color = 'DarkGray';
my $title-color = 'Silver';
```

## Box-Whiscker chart

```raku
#% js
js-d3-box-whisker-chart((1..20).List, :width(500), :horizontal, :$background, :$stroke-color)
==> js-xkcd-modifier(roughness => 0.015)
```

```raku
#%js
random-variate(BetaDistribution.new(120,10), 500) 
==> js-d3-box-whisker-chart(:horizontal, :$background, :$stroke-color) 
==> js-xkcd-modifier(roughness => 0.035)
```

## List-line plot

```raku
%% js
my @data = (
    |((0, 5, 10, 15, 20, 25) Z=> (18, 25, 21, 34, 31, 43)).map(-> $p { %(x => $p.key, y => $p.value, group => 'North') }),
    |((0, 5, 10, 15, 20, 25) Z=> (12, 17, 28, 23, 35, 38)).map(-> $p { %(x => $p.key, y => $p.value, group => 'South') }),
    |((0, 5, 10, 15, 20, 25) Z=> (25, 21, 30, 37, 33, 48)).map(-> $p { %(x => $p.key, y => $p.value, group => 'West') })
);

js-d3-list-line-plot(
    @data,
    width => 900, height => 520,
    title => 'Quarterly progress, approximately',
    :$title-color,
    x-axis-label => 'Week', y-axis-label => 'Score',
    :legends, :grid-lines, stroke-width => 3,
    :$background
) ==> js-xkcd-modifier(roughness => 0.054, wobble => 3.5)
```

## Clocks

```raku
#% js
js-d3-clock-gauge(background => 'none', :$stroke-color)
==> js-xkcd-modifier(roughness => 0.015)
```

```raku
#% js
my @scale-ranges = (0, 0.01 ... 0.66).rotor(2=>-1).map({ ([0, 60], $_) });
js-d3-clock-gauge(background => 'none', 
    hour-hand-color => 'Orange', second-hand-color => 'Red',
    :$stroke-color, 
    fill-color => 'DarkSlateGray', 
    color-scheme => 'Cividis', 
    :@scale-ranges)
==> js-xkcd-modifier(roughness => 0.015, ink-bleed => 0.5, paper-grain => 0, wobble => 10.1)
```
