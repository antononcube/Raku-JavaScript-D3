# 2D density charts: rectangular and hexagonal bins

This notebook demonstrates the two `js-d3-density2d-chart` variants provided by `JavaScript::D3`. Both charts use the same deterministic point set; only the bin geometry changes. The implementation uses D3 v7 scales and data joins, without `d3.rectbin` or `d3.hexbin` plug-ins.

Run the setup cells in order before evaluating either chart cell.

## Load the Raku package

```raku
use JavaScript::D3;
```

## Load D3 v7 in the notebook frontend

The generated visualizations use RequireJS to obtain the `d3` module, so the browser frontend needs this setup once per kernel session.

```raku
%% javascript
require.config({
  paths: { d3: 'https://d3js.org/d3.v7.min' }
});

require(['d3'], function(d3) {
  window.d3 = d3;
  console.log('D3 version:', d3.version);
});
```

## Generate a reproducible two-cluster point set

The values are generated from deterministic trigonometric curves with small repeating offsets. This makes the notebook rerunnable while still producing visibly dense regions.

```raku
my @points;

for ^240 -> $i {
    my $angle = $i * 0.37;
    my $jx = (($i * 17) % 11 - 5) * 0.55;
    my $jy = (($i * 23) % 13 - 6) * 0.45;
    @points.push: %(
        x => 30 + 12 * cos($angle) + $jx,
        y => 38 + 8 * sin($angle) + $jy
    );
}

for ^180 -> $i {
    my $angle = $i * 0.43;
    my $jx = (($i * 19) % 9 - 4) * 0.65;
    my $jy = (($i * 29) % 15 - 7) * 0.40;
    @points.push: %(
        x => 66 + 9 * cos($angle) + $jx,
        y => 61 + 11 * sin($angle) + $jy
    );
}

say "Generated {@points.elems} points.";
```

## Rectangular grid

`method => 'rectangle'` is an alias for `rectbin`. Points are aggregated into a data-dependent rectangular grid, then rendered as SVG rectangles through a D3 data join.

```raku
#% js
js-d3-density2d-chart(
    @points,
    method => 'rectangle',
    width => 760, height => 480,
    title => 'Point density — rectangular bins',
    title-color => 'Silver',
    x-axis-label => 'x coordinate',
    y-axis-label => 'y coordinate',
    grid-lines => (6, 6),
    color => 'steelblue',
    background => 'none'
)
```

## Hexagonal grid

`method => 'hexagon'` is an alias for `hexbin`. Points are aggregated by hex-grid center and rendered as SVG polygons. The radius is measured in plot pixels, so it remains visually stable for the chosen dimensions.

```raku
%% js
js-d3-density2d-chart(
    @points,
    method => 'hexagon',
    width => 760, height => 480,
    title => 'Point density — hexagonal bins',
    title-color => 'Silver',
    x-axis-label => 'x coordinate',
    y-axis-label => 'y coordinate',
    grid-lines => (6, 6),
    color => 'salmon',
    background => 'none'
)
```

## Notes

- Use `rect`, `rectangle`, or `rectbin` for the rectangular variant.
- Use `hex`, `hexagon`, or `hexbin` for the hexagonal variant.
- Both modes accept ordinary numeric coordinate pairs or records with numeric `x` and `y` fields.
- The fill intensity represents the relative count of points in each occupied bin.
