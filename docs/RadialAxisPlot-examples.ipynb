## Setup

```raku
use JavaScript::D3;

use Data::Reshapers;
use Data::Summarizers;
use Data::Generators;
use Data::Translators;
use Data::TypeSystem;
```

The generated visualizations use RequireJS to obtain the `d3` module, so the browser frontend needs this setup once per kernel session.

```raku
%% javascript
require.config({
  paths: { d3: 'https://d3js.org/d3.v7.min' }
});

require(['d3'], function(d3) {
  window.d3 = d3;
  console.log('D3 version:', d3.version);
});
```

## Single dataset

Plot single dataset:

```raku
#% js
my @data = random-real([-1,5], 12);
js-d3-radial-axis-plot(@data, :$background, :$title-color, :3stroke-width, :axes)
```

### Named columns

```raku
my @dsRandom = random-tabular-dataset(20, 6, generators => [{ random-real(10, $_) },] );
@dsRandom = @dsRandom».round(0.01);
deduce-type(@dsRandom)
```

Summarize data:

```raku
sink records-summary(@dsRandom)
```

```raku
#% js
js-d3-radial-axis-plot(
    @dsRandom[12],
    :800width,
    :400height,
    :$background, 
    :$title-color
)
```

## Multiple datasets

Multiple datasets assignment:

```raku
my @md-data = [3,9,9,7,2],[8,10,0,6,7],[7,10,2,4,3];
deduce-type(@md-data)
```

Data summary:

```raku
my @field-names = (^@md-data.&dimensions.tail)».Str;
sink records-summary(@md-data)
```

Multiple datasets plot:

```raku
#% js
@md-data ==>
js-d3-radial-axis-plot(
    :800width,
    :400height, 
    :$background, 
    :$title-color,
    :3stroke-width,
    color-palette => 'Tableau10',
    grid-lines => 'polygon',
    grid-lines-color => '#3B3B3B',
    :axes
)
```

Using long form data:

```raku
my @dsRandom = 
    |random-tabular-dataset(20, [|'ABCD'.comb,'group'], generators => [|({ random-variate(NormalDistribution.new(-1, 0.5), $_) } xx 4), <normal normal>] ),
    |random-tabular-dataset(20, [|'ABCD'.comb,'group'], generators => [|({ 8 <<*>> random-variate(BetaDistribution.new(20, 25), $_) } xx 4), <beta beta>] ),
    |random-tabular-dataset(20, [|'ABCD'.comb,'group'], generators => [|({ random-real([5,9], $_) } xx 4), <uniform uniform>] )
    ;
deduce-type(@dsRandom)
```

```raku
#% js
@dsRandom.head(6) ==>
js-d3-radial-axis-plot(
    :800width,
    :400height, 
    :$background, 
    :$title-color,
    :3stroke-width,
    :filling
)
```
